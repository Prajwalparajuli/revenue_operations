### Exporing the raw data

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Current working directory for the notebook
dir = Path.cwd()

# Navigate to the root directory
root = dir.parent

# Define the absolute path of the file to be opened
data_dir = root / "data" / "raw"

# Read the files in loop and add the names for the files to the dataframes
dataframes = {}

for file_path in data_dir.glob("*.csv"):
    distinct_names = file_path.stem
    dataframes[distinct_names] = pd.read_csv(file_path)

print(dataframes.keys())
```

### Candidate Key test

```python
def scan_candidate_keys(datasets_dict: dict[str, pd.DataFrame]) -> pd.DataFrame:
    results = []
    for datasets_name, df in datasets_dict.items():
        total_rows = len(df)
        for col in df.columns:
            unique_keys = df[col].nunique(dropna=False)
            duplicate_keys = total_rows - unique_keys
            is_unique = "Yes" if duplicate_keys == 0 else "No"

            results.append({
                "Dataset" : datasets_name,
                "Column Name" : col,
                "Total Rows" : total_rows,
                "Unique Keys" : unique_keys,
                "Duplicate Keys" : duplicate_keys,
                "Unique" : is_unique
            })
    return pd.DataFrame(results)
```

```python
datasets = {"Customers" : customers,
            "Geolocation" : geolocation,
            "Ordered Items": ordered_items,
            "Payments" : payments,
            "Reviews" : reviews,
            "Orders" : orders,
            "Products" : products,
            "Sellers" : sellers,
            "Product category translation" : product_category}

report = scan_candidate_keys(datasets)
report[report["Dataset"] == "Customers"]
```
